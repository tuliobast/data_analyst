```python
# para establecer las cabeceras del dataframe se usa el parametro names
df = pd.read_csv('examples/ex2.csv', names=['a','b','c','d', 'message'])
df
```

```python
# manejo de los valores faltante en pandas (NaN)
df = pd.read_csv('examples/ex5.csv')
df
```

```python
# para saber cuales son los valores faltantes usamos isna()
df.isna()
```

```python
# para sustituir cualquier valor por un NaN usamos el parametro na_values
df = pd.read_csv('examples/ex5.csv', na_values=['foo']) # en lugar de NULL podemos colocar cualquier valor para sustituirlo
df
```

```python
# para deshabilidar que pandas lea por defecto los valores nulos parametro keep_default_na
df = pd.read_csv('examples/ex5.csv', keep_default_na=False)
df
```

```python
df.isna()
```

```python
# para diferenciar valores vacios por valores faltantes usamos keep_default_na y na_values
df = pd.read_csv('examples/ex5.csv', keep_default_na=False, na_values=['NA'])
df
```

```python
df.isna()
```

```python
# para establecer varios valores como nulos pasamos un diccionario al parametro na_values
sentinels = {'message': ['foo', 'NA'], 'something': ['two']}
df = pd.read_csv('examples/ex5.csv', na_values=sentinels)
df
```

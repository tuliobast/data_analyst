### **Creando ndarrays**

```python
# creando array a partir de listas
data1 = [6, 7.5, 8, 0, 1]
arr1 = np.array(data1)
print(arr1)

data2 = [[1, 2, 3], [4, 5, 6],[7, 8, 9]]
arr2 = np.array(data2)
print(arr2)
print(arr2.ndim) # imprime las dimenciones del array
print(arr2.shape) # imprime las columnas y filas del array

# array de ceros
print(np.zeros(10))
print(f'zero array(3,6): {np.zeros((3, 6))}')

# array de unos
print(np.ones(10))
print(f'one array(3,6): {np.ones((3, 6))}')
np.ones(())

# array vacio
print(f'empty array: {np.empty((1, 2, 3, 4))}')

# array de un rango
print(f'range array: {np.arange(15)}')

# array de 5
print(f'full array: {np.full((3, 5), 5, dtype=float)}')

# array eye
print(f'eye array: {np.eye(5,2,dtype=str)}')

# array identity
print(f'identity array: {np.identity(6, dtype=bool)}')
```

```python
# para cambiar el tipo de dato contenido en el array astype()
arr = np.array([1, 2, 3, 4, 5])
print(arr.dtype)
float_arr = arr.astype(np.float64)
print(float_arr.dtype)

# los tipos de datos string se declaran con .bytes_
string_arr = np.array(
['1.99', '-9.6', '42'], dtype=np.bytes_)
print(string_arr)
print(string_arr.dtype)
#
```

### **Aritmética con arrays NumPy**
los array en numpy permiten realizar operaciones aritmenticas elemento por elemento si necesidad de un bocle for

```python
arr = np.array([[1., 2., 3.], [4., 5., 6.]])
arr * arr
```

```python
arr - arr
```

```python
# las operaciones con valores escalares se propagan a todo el array
1 / arr
```

####indexado

```python
arr = np.arange(10)
arr[5:8] = 64
arr
# en los arrays los datos no se copian, y que cualquier modificación de la vista se verá reflejada en el array de origen.
arr_slice = arr[5:8]
arr_slice
```

```python
arr_slice[1] = 12
arr
```

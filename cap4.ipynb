### **Creando ndarrays**

```python
# creando array a partir de listas
data1 = [6, 7.5, 8, 0, 1]
arr1 = np.array(data1)
print(arr1)

data2 = [[1, 2, 3], [4, 5, 6],[7, 8, 9]]
arr2 = np.array(data2)
print(arr2)
print(arr2.ndim) # imprime las dimenciones del array
print(arr2.shape) # imprime las columnas y filas del array

# array de ceros
print(np.zeros(10))
print(f'zero array(3,6): {np.zeros((3, 6))}')

# array de unos
print(np.ones(10))
print(f'one array(3,6): {np.ones((3, 6))}')
np.ones(())

# array vacio
print(f'empty array: {np.empty((1, 2, 3, 4))}')

# array de un rango
print(f'range array: {np.arange(15)}')

# array de 5
print(f'full array: {np.full((3, 5), 5, dtype=float)}')

# array eye
print(f'eye array: {np.eye(5,2,dtype=str)}')

# array identity
print(f'identity array: {np.identity(6, dtype=bool)}')
```

```python
# para cambiar el tipo de dato contenido en el array astype()
arr = np.array([1, 2, 3, 4, 5])
print(arr.dtype)
float_arr = arr.astype(np.float64)
print(float_arr.dtype)

# los tipos de datos string se declaran con .bytes_
string_arr = np.array(
['1.99', '-9.6', '42'], dtype=np.bytes_)
print(string_arr)
print(string_arr.dtype)
#
```

####indexado booleano

```python
# array de nombre
names = np.array(["Bob", "Joe", "Jack", "Bob", "Will", "Joe", "Joe"])
# array de datos
data = np.array([[4, 7, 4], [0, 2, 5], [-5, 6, 8], [0, 0, 1], [1, 2, 3], [-12, -4, 5], [3, 4, 7]])
print(names)
print(data)

# queremos relacionar un array de nombres con array de datos
print(f'''data[names == "Jack"]:
{data[names == "Jack"]}
''')

# incluso se pueden seleccionar los datos especificos que se quieren sacar con slicing

print(f''' data[names == "Jack", :1])
{data[names == "Jack", :1]}
''')

# tambien se puede negar la condicion con ~
condition = names == "Jack"
print(f'''data[~condition]
{data[~condition]}
''')

# o usando !=
print(f'''data[names != "Jack"]
{data[names != "Jack"]}
''')

# combinando condiciones
mask = (names == "Jack") | (names == "Will")
print(f'''data[mask]
{data[mask]}
''')

# creado una copia de la lista de datos de jack
jack = data[names == "Jack"]
# para verificar que no se modifican el array de datos
jack[0,0] = 7 # uso [0,0] porque jack es[[]]
print(f'''jack
{jack}
dimencion de jack:
{len(jack.shape)}
''')
print(f'''data[names == "Jack"]
{data[names == "Jack"]}
''')

# para cambiar los valores del array datos mediante condicion
data[data < 0] = 0
print(f'''data[data < 0] = 0
{data}
''')

data[names == "Jack"] = 7
print(f'''data[names == "Jack"] = 7
{data}
''')
```

Seleccionar datos de un array mediante indexado  booleano y asignar el resultado a una nueva variable siempre crea una copia de los datos, incluso aunque el array devuelto no haya sido modificado

####indexado sofisticado

```python
arr = np.zeros((8, 4))
for i in range(8):
  arr[i] = i
arr
```

```python
print(f'''arr[[4, 3, 0, 6]]:
{arr[[4, 3, 0, 6]]}
''')

```

```python
print(f'''arr[[-3, -5, -7]]:
{arr[[-3, -5, -7]]}
''')
```

```python
arr = np.arange(32).reshape((8, 4))
print(arr)
print(f'''
arr[[1, 5, 7, 2], [0, 3, 1, 2]]:
{arr[[1, 5, 7, 2], [0, 3, 1, 2]]}
devuelve
arr[1,0]: {arr[1,0]}
arr[5,3]: {arr[5,3]}
arr[7,1]: {arr[7,1]}
arr[2,2]: {arr[2,2]}''')
```

```python
print(f'''arr[[1, 5, 7, 2]][:, [0, 3, 1, 2]]
{arr[[1, 5, 7, 2]][:, [0, 3, 1, 2]]}
devuelve
arr[[1, 1, 1, 1], [0, 3, 1, 2]]: {arr[[1, 1, 1, 1], [0, 3, 1, 2]]}
arr[[5, 5, 5, 5], [0, 3, 1, 2]]: {arr[[5, 5, 5, 5], [0, 3, 1, 2]]}
arr[[7, 7, 7, 7], [0, 3, 1, 2]]: {arr[[7, 7, 7, 7], [0, 3, 1, 2]]}
arr[[2, 2, 2, 2], [0, 3, 1, 2]]: {arr[[2, 2, 2, 2], [0, 3, 1, 2]]}
''')
```

```python
# para asignar valores con indexado sofisticado
print(arr[[1, 5, 7, 2], [0, 3, 1, 2]])
arr[[1, 5, 7, 2], [0, 3, 1, 2]] = 0
print(arr)
```

####indexado booleano

```python
# array de nombre
names = np.array(["Bob", "Joe", "Jack", "Bob", "Will", "Joe", "Joe"])
# array de datos
data = np.array([[4, 7, 4], [0, 2, 5], [-5, 6, 8], [0, 0, 1], [1, 2, 3], [-12, -4, 5], [3, 4, 7]])
print(names)
print(data)

# queremos relacionar un array de nombres con array de datos
print(f'''data[names == "Jack"]:
{data[names == "Jack"]}
''')

# incluso se pueden seleccionar los datos especificos que se quieren sacar con slicing

print(f''' data[names == "Jack", :1])
{data[names == "Jack", :1]}
''')

# tambien se puede negar la condicion con ~
condition = names == "Jack"
print(f'''data[~condition]
{data[~condition]}
''')

# o usando !=
print(f'''data[names != "Jack"]
{data[names != "Jack"]}
''')

# combinando condiciones
mask = (names == "Jack") | (names == "Will")
print(f'''data[mask]
{data[mask]}
''')

# creado una copia de la lista de datos de jack
jack = data[names == "Jack"]
# para verificar que no se modifican el array de datos
jack[0,0] = 7 # uso [0,0] porque jack es[[]]
print(f'''jack
{jack}
dimencion de jack:
{len(jack.shape)}
''')
print(f'''data[names == "Jack"]
{data[names == "Jack"]}
''')

# para cambiar los valores del array datos mediante condicion
data[data < 0] = 0
print(f'''data[data < 0] = 0
{data}
''')

data[names == "Jack"] = 7
print(f'''data[names == "Jack"] = 7
{data}
''')
```

Seleccionar datos de un array mediante indexado  booleano y asignar el resultado a una nueva variable siempre crea una copia de los datos, incluso aunque el array devuelto no haya sido modificado

###**Expresar lógica condicional como operaciones de arrays**

```python
xarr = np.array([1.1, 1.2, 1.3, 1.4, 1.5])
yarr = np.array([2.1, 2.2, 2.3, 2.4, 2.5])
cond = np.array([True, False, True, True, False])
```

```python
result = [(x if c else y)
  for x, y, c in zip(xarr, yarr, cond)]
print(result)
```

lento y muy poco eficiente

```python
# usando where()
result = np.where(cond, xarr, yarr)
print(result)

# usamos where() para obtener un nuevo array en base a otro con una condicion
arr = rng.standard_normal((4, 4))
print(f'''arr:
{arr}
''')
new_arr = np.where(arr > 0, 2, -1)
print(f'''np.where(arr > 0, 2, -1):
{new_arr}
''')

# combinando escalares y arr con where()
new_arr = np.where(arr < 0, -1, arr)
print(f'''np.where(arr < 0, -1, arr):
{new_arr}
''')
```

##**Ejemplo: de caminos aleatorios**


```python
# forma en python puro de camino aleatorio
#! blockstart
import random
position = 0
walk = [position]
nsteps = 1000
for _ in range(nsteps):
  step = 1 if random.randint(0, 1) else -1
  position += step
  walk.append(position)
#! blockend
```

```python
plt.plot(walk[:100])
```

```python
# ahora con numpy
nsteps = 1000
rng = np.random.default_rng(seed=12345)  # fresh random generator
draws = rng.integers(0, 2, size=nsteps)
steps = np.where(draws == 0, 1, -1)
walk = steps.cumsum()
plt.plot(walk[:100])
```

```python
walk.min()
```

```python
walk.max()
```

```python
(np.abs(walk) >= 10).argmax()
```

###**Simulando muchos caminos aleatorios al mismo tiempo**

```python
nwalks = 5000
nsteps = 1000
draws = rng.integers(0, 2, size=(nwalks, nsteps)) # 0 or 1
steps = np.where(draws > 0, 1, -1)
walks = steps.cumsum(axis=1)
# plt.plot(walks[:20])
```

```python
walks.max()
```

```python
walks.min()
```

```python
# calculamos el tiempo de cruce para 30 0 - 30
hits30 = (np.abs(walks) >= 30).any(axis=1)
print(f'''hits30:
{hits30}
''')

hits30.sum() # Number that hit 30 or -30

# ahora obtenemos de los 5000 los que realmente cruzaron en 30 o -30
crossing_times = (np.abs(walks[hits30]) >= 30).argmax(axis=1)
print(f'''crossing_times:
{crossing_times}
''')
# ahora obtenemos el tiempo promedio minino de cruce
crossing_times.mean()

draws
```
